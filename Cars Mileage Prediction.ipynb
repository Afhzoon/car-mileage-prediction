## 1. Import Necessary Libraries

```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings(action = "ignore")
```

## 2. Import Dataset

```python
cars_data = pd.read_csv(filepath_or_buffer = "Cars.csv")
cars_data
```

### Now let's check for the Assumption

### Test 1 - Linearity Test 

```python
sns.lmplot(data = cars_data, x="HP" , y="MPG")
```

```python
sns.lmplot(data = cars_data, x="VOL" , y="MPG")
```

```python
sns.lmplot(data = cars_data, x="SP" , y="MPG")
```

#### Test 1: Linearity Failed ❌

### Test 2 - Normality Test 

```python
plt.hist(cars_data["HP"])
```

```python
sns.distplot(x = cars_data["HP"])
plt.title("HP Distribution")
plt.show()
```

#### Test 2: Normality Failed ❌

### Test 3 - Multicollinearity Test 

```python
corr_matrix = cars_data.corr()
corr_matrix
```

```python
sns.heatmap(data = corr_matrix, annot = True)
plt.show()
```

#### Test 3: There is Multicollinearity in the Data ❌

#### Test 4: Auto Regression Pass ✅

## 4. Data Preparation

```python
cars_data
```

```python
x = cars_data[["HP","VOL","SP","WT"]]
x
```

```python
y = cars_data["MPG"]
y
```

#### There is nothing to prepare bcz all features are numeric 

## 5. Model Building

```python
linear_model = LinearRegression()
```

## 6. Model Training

```python
linear_model.fit(x,y)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

## 7. Model Testing

```python
y_pred = linear_model.predict(x)
y_pred
```

## 8. Model Evaluation

```python
error = y - y_pred
error
```

```python
cars_data["error"] = error
cars_data
```

#### Let's now perform Test 5: Homoscedasticity Test

```python
plt.scatter(data = cars_data , x = "HP", y = "error")
```

```python
sns.scatterplot(data=cars_data, x="HP", y=error)
```

#### Homoscedasticity Test failed

### Test 6: Zero Residual Mean Test

```python
sns.scatterplot(x = cars_data["MPG"] , y = y_pred )
plt.show()
```

#### Test 6: Zero Residual Mean is Test also Failed

## 9. Model Deployment 

```python
from pickle import dump
```
